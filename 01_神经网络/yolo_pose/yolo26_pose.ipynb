```python
!pip install -q onnx onnxruntime onnxslim

from ultralytics import YOLO
import onnxruntime as ort

best_pt = "/content/runs/pose/yolo26n-full/weights/best.pt"

model = YOLO(best_pt)

onnx_path = model.export(
    format="onnx",
    imgsz=640,
    opset=12,
    simplify=True,
    dynamic=False,
    nms=True,
    device="cpu"
)

print("导出完成：", onnx_path)

session = ort.InferenceSession(
    str(onnx_path),
    providers=["CPUExecutionProvider"]
)

for output in session.get_outputs():
    print("输出：", output.name, output.shape)
```
